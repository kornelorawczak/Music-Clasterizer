# Wizualizacja reprezentacji w różnych formach
Kluczowe jest aby skalować wszystkie cechy do zakresu 0-1, ponieważ każda cecha ma totalnie inne przedziały

````python
from sklearn.manifold import TSNE
import pandas as pd
import altair as alt
from sklearn.decomposition import PCA
import umap
from sklearn.preprocessing import StandardScaler

df_scaled = df.copy()
df_librosa_scaled = df_librosa.copy()

scaler1 = StandardScaler()
scaler2 = StandardScaler()
df_scaled = scaler1.fit_transform(df_scaled)
df_librosa_scaled = scaler2.fit_transform(df_librosa_scaled)

tsne = TSNE(n_components=2, random_state=42, perplexity=5)
X_tsne = tsne.fit_transform(df_scaled)
X_tsne_librosa = tsne.fit_transform(df_librosa_scaled)

pca = PCA(n_components=2, random_state=42)
X_pca = pca.fit_transform(df_scaled)
X_pca_librosa = pca.fit_transform(df_librosa_scaled)

ump = umap.UMAP(n_neighbors=8, min_dist=0.1, n_components=2, metric='cosine', random_state=42)
X_umap = ump.fit_transform(df_scaled)
X_umap_librosa = ump.fit_transform(df_librosa_scaled)

def create_scatter(X, id, title, titles, albums):
    df_vis = pd.DataFrame({
        'x': X[:, 0],
        'y': X[:, 1],
        'Title': titles,
        'Album': albums
    })

    df_centroids = df_vis.groupby('Album')[['x', 'y']].mean().reset_index()
    df_centroids['Title'] = df_centroids['Album'].apply(lambda x: f"AVERAGE of {x}")
    df_centroids['Type'] = 'Centroid'

    songs = alt.Chart(df_vis).mark_circle(size=60).encode(
        x=alt.X('x', title='Feature 1'),
        y=alt.Y('y', title='Feaeture 2'),
        color=alt.Color('Album', title='Album Name'), 
        tooltip=['Title', 'Album'] 
    )

    centers = alt.Chart(df_centroids).mark_point(
        shape='diamond', 
        size=150,
        filled=True, 
        opacity=0.7,
        stroke='black', 
        strokeWidth=2
    ).encode(
        x='x',
        y='y',
        color='Album',
        tooltip=['Title', 'Album'] 
    )

    chart = (songs + centers).properties(
        title=title,
        width=700,
        height=600
    ).interactive()

    chart.save(f'scatter_plot_{id}.html')
    chart.show()

create_scatter(X_tsne, "tsne", "t-SNE 2D visualization", titles, albums)
create_scatter(X_tsne_librosa, "tsne_librosa", "t-SNE 2D visualization using librosa representation", titles_librosa, albums_librosa)
create_scatter(X_pca, "pca", "PCA 2D visualization", titles, albums)
create_scatter(X_pca_librosa, "pca_librosa", "PCA 2D visualization using librosa", titles_librosa, albums_librosa)
create_scatter(X_umap, "umap", "UMAP 2D visualization", titles, albums)
create_scatter(X_umap_librosa, "umap_librosa", "UMAP 2D visualization using librosa", titles_librosa, albums_librosa)

````

# Analyzing the original space

````python
import numpy as np
from scipy.spatial.distance import cdist, pdist, squareform

def analyze_kanye_discography(X, song_titles):
    centroid = np.mean(X, axis=0).reshape(1, -1)
    distances_to_mean = cdist(X, centroid, metric='euclidean').flatten()
    
    # Get top 3 indices with smallest distance
    typical_indices = np.argsort(distances_to_mean)[:5]
    print("\n--- Top 5 Most Typical Songs (Closest to Average Sound) ---")
    for idx in typical_indices:
        print(f"{song_titles[idx]} (Distance to mean: {distances_to_mean[idx]:.4f})")

    internal_stds = np.std(X, axis=1)
    experimental_indices_std = np.argsort(internal_stds)[::-1][:5]
    
    print("\n--- Top 5 Experimental Songs (Highest Feature Variance) ---")
    for idx in experimental_indices_std:
        print(f"{song_titles[idx]} (Std Dev: {internal_stds[idx]:.4f})")

    # Furthest from the Mean
    outlier_indices = np.argsort(distances_to_mean)[::-1][:5]
    print("\n--- Top 5 Outlier Songs (Furthest from Average Sound) ---")
    for idx in outlier_indices:
        print(f"{song_titles[idx]} (Distance from mean: {distances_to_mean[idx]:.4f})")

    # Calculate distance between every pair of songs
    dist_matrix = squareform(pdist(X, metric='euclidean'))
    
    # Set diagonal to infinity so a song isn't its own closest neighbor
    np.fill_diagonal(dist_matrix, np.inf)
    
    # Find the pairs with the absolute minimum distances
    # We flatten the upper triangle to avoid duplicates (A-B and B-A)
    # Create a list of (song_i, song_j, distance)
    pairs = []
    n_songs = len(song_titles)
    for i in range(n_songs):
        for j in range(i + 1, n_songs):
            pairs.append((song_titles[i], song_titles[j], dist_matrix[i, j]))

    # Sort by distance (ascending)
    pairs.sort(key=lambda x: x[2])
    
    print("\n--- Top 10 Closest Pairs (Most Similar Sounding) ---")
    for s1, s2, d in pairs[:10]:
        print(f"- {s1}  <-->  {s2} (Distance: {d:.4f})")

print("ANALYZING ORIGINAL SPACE USING CUSTOM REPRESENTATION\n")
analyze_kanye_discography(df_scaled, titles)
print("\nANALYZING ORIGINAL SPACE USING LIBROSA REPRESENTATION\n")
analyze_kanye_discography(df_librosa_scaled, titles_librosa)
````

# Tuning UMAP Parameters

````python
from sklearn.cluster import HDBSCAN
from sklearn.metrics import silhouette_score
from itertools import product
import warnings
warnings.filterwarnings('ignore')

def tune_umap_params(X):
    umap_neighbors = [3, 5, 8, 10, 12, 15, 20, 24, 30, 35, 40, 50]
    umap_min_dist = [0.0, 0.05, 0.1, 0.25, 0.50, 0.75, 0.85, 0.99]

    hdb_min_cluster_size = [3, 4, 5, 7, 10, 15]
    hdb_min_samples = [1, 2, 3, 5, 8, 12]

    results = []

    for n, min_dist in product(umap_neighbors, umap_min_dist):
        ump = umap.UMAP(n_neighbors=n, min_dist=min_dist, n_components=2, random_state=42, n_jobs=-1)
        X_2d = ump.fit_transform(X)

        for min_c, min_s in product(hdb_min_cluster_size, hdb_min_samples):
            # Chce więcej klastrów więc zmieniam metode klastrowania na liściastą
            clusterer = HDBSCAN(min_cluster_size=min_c, min_samples=min_s, cluster_selection_method='leaf')
            labels = clusterer.fit_predict(X_2d)

            # Nie chcemy brać pod uwagę szumu (label = -1) w liczeniu score
            clustered_mask = labels != -1
            n_clusters = len(set(labels)) - (1 if -1 in labels else 0)

            # Jeśli wszystko jest szumem albo jest tylko jeden klaster to nas to nie interesuje
            if n_clusters < 2 or np.sum(clustered_mask) < len(labels) * 0.5:
                continue

            score_2d = silhouette_score(X_2d[clustered_mask], labels[clustered_mask])
            score_original = silhouette_score(X[clustered_mask], labels[clustered_mask])
            noise_ratio = np.sum(labels == -1) / len(labels)

            # Lekko promujemy rozwiązania, które mają więcej klastrów 
            adjusted_score = score_original + (0.008 * n_clusters)

            results.append({
                'n_neighbors': n,
                'min_dist': min_dist,
                'min_cluster_size': min_c,
                'min_samples': min_s,
                'score_original': score_original,
                'adjusted_score': adjusted_score, 
                'score_2d': score_2d,
                'noise_ratio': noise_ratio,
                'n_clusters': n_clusters,
                'embedding': X_2d,
                'labels': labels
            })

    df_res = pd.DataFrame(results)
    df_res = df_res[df_res['noise_ratio'] < 0.35]
    best_result = df_res.sort_values(by='adjusted_score', ascending=False).iloc[0]
    # best_result = df_res.sort_values(by='score_2d', ascending=False).iloc[0]

    print("\n--- WYNIKI OPTYMALIZACJI HDBSCAN ---")
    print(f"Params: UMAP(n={best_result['n_neighbors']}, min_dist={best_result['min_dist']}) \n HDBSCAN(min_cluster_size={best_result['min_cluster_size']}, min_samples={best_result['min_samples']})")
    print(f"Liczba wykrytych klastrów: {int(best_result['n_clusters'])}")
    print(f"Odrzucone utwory (Szum): {best_result['noise_ratio']:.2%}")
    print(f"Jakość Silhouette w og przestrzeni: {best_result['score_original']:.4f}")
    print(f"Jakość Silhouette w 2d przestrzeni: {best_result['score_2d']:.4f}")

    return best_result['embedding'], best_result['labels'], best_result['n_clusters']

def create_scatter_hdbscan(X, labels, id, title, titles, albums):
    cluster_names = []
    for l in labels:
        if l == -1:
            cluster_names.append("Noise (Unclassified)")
        else:
            cluster_names.append(f"Cluster {l}")
            
    df_vis = pd.DataFrame({
        'x': X[:, 0],
        'y': X[:, 1],
        'Title': titles,
        'Original_Album': albums,
        'Cluster': cluster_names,
        'IsNoise': [1 if l == -1 else 0 for l in labels]
    })

    # Centroidy liczymy tylko dla poprawnych klastrów (bez szumu)
    df_centroids = df_vis[df_vis['IsNoise'] == 0].groupby('Cluster')[['x', 'y']].mean().reset_index()
    df_centroids['Title'] = df_centroids['Cluster'].apply(lambda x: f"CENTER of {x}")

    songs = alt.Chart(df_vis).mark_circle(size=60).encode(
        x=alt.X('x', title='UMAP 1'),
        y=alt.Y('y', title='UMAP 2'),
        color=alt.condition(
            alt.datum.Cluster == 'Noise (Unclassified)',
            alt.value('lightgray'), 
            alt.Color('Cluster', scale=alt.Scale(scheme='tableau20'), title='Sound Cluster') 
        ),
        tooltip=['Title', 'Original_Album', 'Cluster'] 
    ).interactive()

    centers = alt.Chart(df_centroids).mark_point(
        shape='cross', 
        size=200,
        filled=True, 
        color='black'
    ).encode(
        x='x',
        y='y',
        tooltip=['Title']
    )
    
    chart = (songs + centers).properties(
        title=title,
        width=750,
        height=600
    )

    filename = f'scatter_hdbscan_{id}.html'
    chart.save(filename)
    chart.show() 


best_embedding, best_labels, best_k = tune_umap_params(df_scaled)
create_scatter_hdbscan(
        best_embedding,
        best_labels,
        id="sound_groups",
        title=f"HDBSCAN Audio Clustering (Detected {best_k} distinct styles)",
        titles=titles,
        albums=albums
    )
````

````python
best_embedding, best_labels, best_k = tune_umap_params(df_librosa_scaled)
create_scatter_hdbscan(
        best_embedding,
        best_labels,
        id="sound_groups",
        title=f"HDBSCAN Audio Clustering (Detected {best_k} distinct styles) for librosa representation",
        titles=titles,
        albums=albums
    )
````
