# Wizualizacja reprezentacji w różnych formach

````python
from sklearn.manifold import TSNE
import pandas as pd
import altair as alt
from sklearn.decomposition import PCA
import umap

tsne = TSNE(n_components=2, random_state=42, perplexity=5)
X_tsne = tsne.fit_transform(df)
X_tsne_librosa = tsne.fit_transform(df_librosa)

pca = PCA(n_components=2, random_state=42)
X_pca = pca.fit_transform(df)
X_pca_librosa = pca.fit_transform(df_librosa)

ump = umap.UMAP(n_neighbors=8, min_dist=0.1, n_components=2, metric='cosine', random_state=42)
X_umap = ump.fit_transform(df)
X_umap_librosa = ump.fit_transform(df_librosa)

def create_scatter(X, id, title, titles, albums):
    df_vis = pd.DataFrame({
        'x': X[:, 0],
        'y': X[:, 1],
        'Title': titles,
        'Album': albums
    })

    df_centroids = df_vis.groupby('Album')[['x', 'y']].mean().reset_index()
    df_centroids['Title'] = df_centroids['Album'].apply(lambda x: f"AVERAGE of {x}")
    df_centroids['Type'] = 'Centroid'

    songs = alt.Chart(df_vis).mark_circle(size=60).encode(
        x=alt.X('x', title='Feature 1'),
        y=alt.Y('y', title='Feaeture 2'),
        color=alt.Color('Album', title='Album Name'), 
        tooltip=['Title', 'Album'] 
    )

    centers = alt.Chart(df_centroids).mark_point(
        shape='diamond', 
        size=150,
        filled=True, 
        opacity=0.7,
        stroke='black', 
        strokeWidth=2
    ).encode(
        x='x',
        y='y',
        color='Album',
        tooltip=['Title', 'Album'] 
    )

    chart = (songs + centers).properties(
        title=title,
        width=700,
        height=600
    ).interactive()

    chart.save(f'scatter_plot_{id}.html')
    chart.show()

create_scatter(X_tsne, "tsne", "t-SNE 2D visualization", titles, albums)
create_scatter(X_tsne_librosa, "tsne_librosa", "t-SNE 2D visualization using librosa representation", titles, albums)
create_scatter(X_pca, "pca", "PCA 2D visualization", titles, albums)
create_scatter(X_pca_librosa, "pca_librosa", "PCA 2D visualization using librosa", titles, albums)
create_scatter(X_umap, "umap", "UMAP 2D visualization", titles, albums)
create_scatter(X_umap_librosa, "umap_librosa", "UMAP 2D visualization using librosa", titles, albums)

````

# Analyzing the original space

````python
import numpy as np
from scipy.spatial.distance import cdist, pdist, squareform

def analyze_kanye_discography(X, song_titles):
    centroid = np.mean(X, axis=0).reshape(1, -1)
    distances_to_mean = cdist(X, centroid, metric='euclidean').flatten()
    
    # Get top 3 indices with smallest distance
    typical_indices = np.argsort(distances_to_mean)[:5]
    print("\n--- Top 5 Most Typical Songs (Closest to Average Sound) ---")
    for idx in typical_indices:
        print(f"{song_titles[idx]} (Distance to mean: {distances_to_mean[idx]:.4f})")

    internal_stds = np.std(X, axis=1)
    experimental_indices_std = np.argsort(internal_stds)[::-1][:5]
    
    print("\n--- Top 5 Experimental Songs (Highest Feature Variance) ---")
    for idx in experimental_indices_std:
        print(f"{song_titles[idx]} (Std Dev: {internal_stds[idx]:.4f})")

    # Furthest from the Mean
    outlier_indices = np.argsort(distances_to_mean)[::-1][:5]
    print("\n--- Top 5 Outlier Songs (Furthest from Average Sound) ---")
    for idx in outlier_indices:
        print(f"{song_titles[idx]} (Distance from mean: {distances_to_mean[idx]:.4f})")

    # Calculate distance between every pair of songs
    dist_matrix = squareform(pdist(X, metric='euclidean'))
    
    # Set diagonal to infinity so a song isn't its own closest neighbor
    np.fill_diagonal(dist_matrix, np.inf)
    
    # Find the pairs with the absolute minimum distances
    # We flatten the upper triangle to avoid duplicates (A-B and B-A)
    # Create a list of (song_i, song_j, distance)
    pairs = []
    n_songs = len(song_titles)
    for i in range(n_songs):
        for j in range(i + 1, n_songs):
            pairs.append((song_titles[i], song_titles[j], dist_matrix[i, j]))

    # Sort by distance (ascending)
    pairs.sort(key=lambda x: x[2])
    
    print("\n--- Top 5 Closest Pairs (Most Similar Sounding) ---")
    for s1, s2, d in pairs[:5]:
        print(f"- {s1}  <-->  {s2} (Distance: {d:.4f})")

analyze_kanye_discography(df, titles)
````
